Let's import necessary modules

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

Loading the csv file into a pandas dataframe

```python
dataset_path = "../../DataSource/Social_media_impact_on_life.csv"
df = pd.read_csv(dataset_path)
df.head()
```

## Univariate Analysis

```python
def do_univariate_analysis(x_column: str, title: str, x_label: str, y_label: str = "Frequency", fig_size: tuple = (8,5)) -> None:

    plt.figure(figsize= fig_size)

    sns.histplot(data= df, x= x_column, color= "skyblue")

    plt.title(title)
    plt.xlabel(xlabel= x_label)
    plt.ylabel(ylabel= y_label)

    plt.show()
```

### Age

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
sns.histplot(data=df, x="Age", color="skyblue", ax=axes[0])
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Frequency")
axes[0].set_title("Distribution of Age")

# Boxplot
sns.boxplot(data=df, x="Age", ax=axes[1])
axes[1].set_xlabel("Age")
axes[1].set_title("Boxplot of Age")

plt.tight_layout()
plt.show()
```

### Avg_Daily_Usage_Hours

```python
plt.figure(figsize= (8,5))

sns.histplot(data= df, x= "Avg_Daily_Usage_Hours", color= "skyblue")

plt.title("Distribution of Average Daily Usage")
plt.xlabel("Average Daily Usage in Hours")
plt.ylabel("Frequency")

plt.show()
```

### Most_Used_Platform

```python
do_univariate_analysis(x_column= "Most_Used_Platform", title= "Distribution of Social Media Platforms", x_label= "Social Media Platforms", fig_size= (14, 6))
```
